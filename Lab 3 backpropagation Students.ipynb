<p><p style="text-align: left;font-size: 1.15em;">Here I have repurposed our perceptron from lab 2 to be used as a neuron in a hidden layer. Perceptron 2 is is the output layer neuron calulation. This is very similar to the summation function back in Lab 2 but this time with addition instead of multiplication. </p></p>

```python
def sigmoid(x):
   # print("why is that happening:"+str(x))
    #print("why is that happening negative edition:"+str(-x))
    return  1/(1+math.exp(-x))
```

```python
def Perceptron(input1,input2,w,b): # this is a neuron in the hidden layer
    return  input1*w[0]+input2*w[1]+b
```

```python
def Perceptron2(inputs,w,b):# this is a neuron in the output layer
    outputP = 0
    for k in range(0,len(weights[len(weights)-1])):
        temp = inputs[k]* w[k]
        outputP += temp
    return sigmoid(outputP + b)
```

```python
def feed_forward(inputs): # inputs should be of form [x,y]
    O = []
    I = [0] * 5
    for k in range(0,nHN):
        temp=sigmoid(Perceptron(inputs[0],inputs[1],weights[k],bias))
        I[k] = temp #this is input from the h????? layer
        
    outPut = Perceptron2(I,weights[nHN],bias2)
    return outPut,I
```

```python
def train(into,out):
    shot,I = feed_forward(into)
    #1 Output neuron deltas
    outputErrorDerivative = (out - shot)
    outputSigmoidDerivative = shot * (1 - shot)
    outputSumDerivative = outputErrorDerivative * outputSigmoidDerivative
    #NOTE: if you do use -(tⱼ - yⱼ) you may have to change the bias' operators around too (eg: += to -=)
    
    global bias2 #you will need to modify the biases
    global bias
    bias2 += outputSumDerivative * lr
    
    #2 Hidden neuron deltas
    #∂zⱼ/∂ = (input(1-input)) COULD THIS HAVE TO DO WITH THE '???' IN FEED_FORWARD???? 
    #wᵢⱼ is the weight from the output layer note the IJ this means output layer while i means hidden layer.
    hiddenSumDerivatives = []
    for i in range(len(weights[nHN])):
        hiddenErrorDerivative = outputSumDerivative * weights[nHN][i]
        hiddenSigmoidDerivative = I[i] * (1 - I[i])
        hiddenSumDerivatives.append(hiddenErrorDerivative * hiddenSigmoidDerivative)
    bias += sum(hiddenSumDerivatives) * lr
    
    #3 Update output neuron weights
    for i in range(nHN):
        weights[nHN][i] += lr * outputSumDerivative * I[i] 
    
    # ∂zⱼ/∂wᵢⱼ  = input from '???' in feedfoward
    #Δw is change in weights and each neuron has multiple neurons
    #α is the learning rate
    
    #4 Update hidden neuron weights
    for i in range(nHN):
        weights[i][0] += lr * hiddenSumDerivatives[i] * into[0]
        weights[i][1] += lr * hiddenSumDerivatives[i] * into[1]
   
    #so if ∂zⱼ/∂wᵢⱼ are the values going into the output layer from the hidden layer then ∂zⱼ/∂wᵢ is input coming form the ????? layer into the ?????? layer
    
    #consider using: https://mattmazur.com/2015/03/17/a-step-by-step-backpropagation-example/
```

```python
def calculate_total_error(training_sets):
    total_error = 0
    for t in range(len(training_sets)):
        training_inputs, training_outputs = training_sets[t]
        out,hidden = feed_forward(training_inputs)
        for o in range(len(training_outputs)):
            total_error += 0.5 * (training_outputs[0] - out) ** 2
            
    return total_error
```
